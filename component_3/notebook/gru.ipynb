```python
# -*- coding: utf-8 -*-
"""
GRU training script -- updated to match lstm_training.py's approach
════════════════════════════════════════════════════════════════
This mirrors the reasoning already applied to the LSTM script. Only the
recurrent layer type differs (GRU instead of LSTM); the imbalance-handling
and checkpointing logic is identical, for the same reasons:

CHANGED vs. the previous model_gru.py:
  - Dropped: focal loss + WEIGHT_SMOOTHING + MANUAL_WEIGHT_OVERRIDE +
    oversampling-with-jitter. That was four interacting hand-tuned knobs,
    arrived at by nudging numbers until validation metrics looked better
    (the script's own comment: naive combination "tanked Waiting recall").
    With only a handful of Hesitant examples in val, that tuning process
    is very likely fitting noise rather than a real effect.
  - Replaced with ONE mechanism: effective number of samples reweighting
    (Cui et al., CVPR 2019) via class_weight= on plain sparse categorical
    crossentropy. One citable formula instead of four hand-tuned constants.
  - Dropped oversampling-with-jitter entirely -- if effective-number
    weighting alone isn't enough, that's a controlled follow-up ablation
    to run and report, not something to bake in from the start.
  - Checkpointing now tracks a short moving average (last 3 epochs) of
    val macro F1, not val_loss (which previously restored an underfit
    early epoch) and not the raw per-epoch macro F1 either (too noisy
    given how few Hesitant/Aggressive examples are in val).

UNCHANGED:
  - Architecture is still two-layer recurrent + dropout + dense, just
    with GRU cells instead of LSTM cells -- the imbalance problem was in
    training/evaluation, not model capacity, so there's no reason to
    touch layer sizes, regularization, or dropout here.
  - Data paths, seeds, batch size, epochs, patience.
════════════════════════════════════════════════════════════════
"""

from google.colab import drive
drive.mount('/content/drive')

# Verify GPU is available
import subprocess
result = subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'],
                        capture_output=True, text=True)
print('GPU:', result.stdout.strip() if result.returncode == 0 else 'Not available — will use CPU')

subprocess.run(['pip', 'install', 'seaborn', '-q'])
print('Ready.')

import os
import json
from collections import deque

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks
from sklearn.metrics import (
    classification_report, confusion_matrix,
    f1_score, accuracy_score
)

print(f'TensorFlow version : {tf.__version__}')
print(f'GPU available      : {len(tf.config.list_physical_devices("GPU")) > 0}')

# ── PATHS ─────────────────────────────────────────────────────
PIE_PATH     = '/content/drive/MyDrive/PIE'
FEATURE_PATH = f'{PIE_PATH}/processed_features_new/processed_sequences'
MODEL_PATH   = f'{PIE_PATH}/processed_features_new/models'
os.makedirs(MODEL_PATH, exist_ok=True)

# ── CONFIG ────────────────────────────────────────────────────
SEQ_LEN    = 30
N_FEATURES = 36
N_CLASSES  = 6
BATCH_SIZE = 32
EPOCHS     = 100
PATIENCE   = 15
SEED       = 19

EFFECTIVE_NUM_BETA = 0.999   # Cui et al. 2019 -- higher beta = gentler
                             # correction for larger datasets. 0.999 is
                             # the standard default; lower (e.g. 0.99)
                             # would weight rare classes more aggressively
                             # if you find this isn't enough.
F1_SMOOTHING_WINDOW = 3      # epochs averaged for checkpoint/stop decisions

LOW_SUPPORT_THRESHOLD = 10   # matches the sequence-building script's own
                             # warning threshold, for consistent reporting

tf.random.set_seed(SEED)
np.random.seed(SEED)

# ── LABEL NAMES ───────────────────────────────────────────────
LABEL_NAMES = [
    'Waiting',     # 0
    'Hesitant',    # 1
    'Committed',   # 2
    'Distracted',  # 3
    'Aggressive',  # 4
    'Jaywalk'      # 5
]

print(f'\nFeature path : {FEATURE_PATH}')
print(f'Model path   : {MODEL_PATH}')

# ============================================================
# 1. LOAD DATA
# ============================================================

X_train = np.load(f'{FEATURE_PATH}/X_train.npy')
y_train = np.load(f'{FEATURE_PATH}/y_train.npy')

X_val = np.load(f'{FEATURE_PATH}/X_val.npy')
y_val = np.load(f'{FEATURE_PATH}/y_val.npy')

X_test = np.load(f'{FEATURE_PATH}/X_test.npy')
y_test = np.load(f'{FEATURE_PATH}/y_test.npy')

print('Data loaded:')
print(f'  X_train : {X_train.shape}  y_train : {y_train.shape}')
print(f'  X_val   : {X_val.shape}   y_val   : {y_val.shape}')
print(f'  X_test  : {X_test.shape}  y_test  : {y_test.shape}')
print(f'  dtype   : {X_train.dtype}')

assert not np.isnan(X_train).any(), 'NaN found in X_train!'
assert not np.isnan(X_val).any(),   'NaN found in X_val!'
assert not np.isnan(X_test).any(),  'NaN found in X_test!'
print('\nNo NaN values detected. Data is clean.')

print('\nClass distribution in training set:')
for i, name in enumerate(LABEL_NAMES):
    n   = int(np.sum(y_train == i))
    pct = n / len(y_train) * 100
    bar = '█' * int(pct / 2)
    print(f'  {i} {name:<15} {n:>5,}  ({pct:5.1f}%)  {bar}')

# Low-support warning for val/test.
print('\nLow-support warning (val/test):')
any_warned = False
for split_name, y_split in [('val', y_val), ('test', y_test)]:
    for lbl in range(N_CLASSES):
        n = int(np.sum(y_split == lbl))
        if n < LOW_SUPPORT_THRESHOLD:
            print(f'  {split_name}: {LABEL_NAMES[lbl]} has only {n} example(s) -- '
                  f'treat this class\'s precision/recall as high-variance.')
            any_warned = True
if not any_warned:
    print(f'  none -- every class has >= {LOW_SUPPORT_THRESHOLD} examples in val and test.')

# ============================================================
# 2. EFFECTIVE-NUMBER-OF-SAMPLES CLASS WEIGHTS
#    Cui et al., "Class-Balanced Loss Based on Effective Number of
#    Samples", CVPR 2019.
# ============================================================

def effective_number_weights(y, n_classes, beta=EFFECTIVE_NUM_BETA):
    counts = {c: int(np.sum(y == c)) for c in range(n_classes)}
    weights = {}
    for c, n in counts.items():
        n = max(n, 1)  # guard against a class with zero examples
        eff_num = (1 - beta**n) / (1 - beta)
        weights[c] = 1.0 / eff_num
    mean_w = sum(weights.values()) / len(weights)
    return {c: w / mean_w for c, w in weights.items()}, counts


class_weights, train_counts = effective_number_weights(y_train, N_CLASSES)

print(f'\nClass weights (effective number of samples, beta={EFFECTIVE_NUM_BETA}):')
for i, name in enumerate(LABEL_NAMES):
    print(f'  {i} {name:<15}  n={train_counts[i]:>5,}  weight={class_weights[i]:.3f}')

# ============================================================
# 3. MODEL (GRU)
# ============================================================

def build_gru_model(seq_len=30, n_features=36, n_classes=6):
    """
    Two-layer GRU for pedestrian mental state classification.
    Same shape/regularization as the LSTM version -- the imbalance
    problem was in training/evaluation, not model capacity, so there's
    no reason to change layer sizes or dropout when swapping cell type.
    """
    inputs = keras.Input(shape=(seq_len, n_features), name='sequence_input')

    x = layers.GRU(
        128, return_sequences=True,
        kernel_regularizer=keras.regularizers.l2(1e-4),
        name='gru_1'
    )(inputs)
    x = layers.Dropout(0.3, name='dropout_1')(x)

    x = layers.GRU(
        64, return_sequences=False,
        kernel_regularizer=keras.regularizers.l2(1e-4),
        name='gru_2'
    )(x)
    x = layers.Dropout(0.3, name='dropout_2')(x)

    outputs = layers.Dense(n_classes, activation='softmax', name='mental_state_output')(x)

    return keras.Model(inputs=inputs, outputs=outputs, name='GRU_MentalState')


model = build_gru_model(SEQ_LEN, N_FEATURES, N_CLASSES)

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-3),
    loss='sparse_categorical_crossentropy',   # standard loss -- class_weight=
    metrics=['accuracy']                       # below does the reweighting
)

model.summary()
print(f'\nTotal parameters : {model.count_params():,}')

# ============================================================
# 4. SMOOTHED MACRO-F1 CHECKPOINT
#    Checkpoints/stops on a moving average of val macro F1, not val_loss
#    (val_loss previously picked an underfit early-epoch model) and not
#    the raw per-epoch macro F1 either (too noisy given 3-7 example
#    classes).
# ============================================================

class SmoothedMacroF1Checkpoint(callbacks.Callback):
    def __init__(self, X_val, y_val, save_path, patience=15, window=3, label_names=None):
        super().__init__()
        self.X_val, self.y_val = X_val, y_val
        self.save_path = save_path
        self.patience = patience
        self.history = deque(maxlen=window)
        self.best_smoothed = -1.0
        self.best_epoch = -1
        self.best_raw_f1 = -1.0
        self.wait = 0
        self.label_names = label_names or [str(i) for i in range(N_CLASSES)]

    def on_epoch_end(self, epoch, logs=None):
        probs = self.model.predict(self.X_val, verbose=0)
        preds = np.argmax(probs, axis=1)
        raw_f1 = f1_score(self.y_val, preds, average='macro', zero_division=0)

        self.history.append(raw_f1)
        smoothed = sum(self.history) / len(self.history)

        print(f'  val_macro_f1={raw_f1:.4f}  smoothed(last {len(self.history)})={smoothed:.4f}'
              f'  (best smoothed so far: {self.best_smoothed:.4f} @ epoch {self.best_epoch})')

        if smoothed > self.best_smoothed:
            self.best_smoothed = smoothed
            self.best_raw_f1 = raw_f1
            self.best_epoch = epoch + 1
            self.wait = 0
            self.model.save(self.save_path)
            print('  -> new best smoothed val macro F1, checkpoint saved.')
        else:
            self.wait += 1
            if self.wait >= self.patience:
                print(f'  -> no improvement in {self.patience} epochs, stopping.')
                self.model.stop_training = True


MODEL_SAVE_PATH = f'{MODEL_PATH}/gru_mental_state_best.keras'

f1_checkpoint = SmoothedMacroF1Checkpoint(
    X_val, y_val, MODEL_SAVE_PATH,
    patience=PATIENCE, window=F1_SMOOTHING_WINDOW, label_names=LABEL_NAMES
)

cb_list = [
    f1_checkpoint,
    callbacks.ReduceLROnPlateau(
        monitor='val_loss', factor=0.5, patience=7, min_lr=1e-6, verbose=1
    ),
    callbacks.CSVLogger(f'{MODEL_PATH}/gru_training_log.csv', append=False)
]

print('Callbacks configured:')
print(f'  SmoothedMacroF1Checkpoint -> {MODEL_SAVE_PATH}')
print(f'                               (window={F1_SMOOTHING_WINDOW}, patience={PATIENCE})')
print(f'  ReduceLROnPlateau  factor=0.5, patience=7')
print(f'  CSVLogger          -> {MODEL_PATH}/gru_training_log.csv')

print('Starting training...')
print(f'  Input shape  : {X_train.shape}')
print(f'  Batch size   : {BATCH_SIZE}')
print(f'  Max epochs   : {EPOCHS}')
print(f'  Class weights: effective-number-of-samples (beta={EFFECTIVE_NUM_BETA})')
print()

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    class_weight=class_weights,
    callbacks=cb_list,
    verbose=1
)

print(f'\nTraining finished at epoch {len(history.history["loss"])}')
print(f'Best smoothed val macro F1: {f1_checkpoint.best_smoothed:.4f}  '
      f'(raw: {f1_checkpoint.best_raw_f1:.4f})  @ epoch {f1_checkpoint.best_epoch}')

# ============================================================
# 5. TRAINING CURVES
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
epochs_ran = range(1, len(history.history['loss']) + 1)

axes[0].plot(epochs_ran, history.history['loss'],     label='Train loss', color='#2563EB', linewidth=2)
axes[0].plot(epochs_ran, history.history['val_loss'], label='Val loss',   color='#DC2626', linewidth=2, linestyle='--')
axes[0].axvline(x=f1_checkpoint.best_epoch, color='gray', linestyle=':', alpha=0.7,
                 label=f'Best epoch {f1_checkpoint.best_epoch} (smoothed macro F1)')
axes[0].set_title('Loss', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Categorical Crossentropy')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

axes[1].plot(epochs_ran, history.history['accuracy'],     label='Train acc', color='#2563EB', linewidth=2)
axes[1].plot(epochs_ran, history.history['val_accuracy'], label='Val acc',   color='#DC2626', linewidth=2, linestyle='--')
axes[1].axhline(y=0.85, color='green', linestyle='--', alpha=0.7, label='Target 85%')
axes[1].set_title('Accuracy', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Accuracy')
axes[1].set_ylim([0, 1])
axes[1].legend()
axes[1].grid(True, alpha=0.3)

plt.suptitle('GRU Mental State Classifier — Training History', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{MODEL_PATH}/gru_training_curves.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved: {MODEL_PATH}/gru_training_curves.png')

# ============================================================
# 6. TEST SET EVALUATION
# ============================================================

best_model = keras.models.load_model(MODEL_SAVE_PATH, compile=False)
print(f'Loaded best model from: {MODEL_SAVE_PATH}  (epoch {f1_checkpoint.best_epoch})')

y_prob = best_model.predict(X_test, batch_size=BATCH_SIZE, verbose=0)
y_pred = np.argmax(y_prob, axis=1)

test_acc  = accuracy_score(y_test, y_pred)
macro_f1  = f1_score(y_test, y_pred, average='macro',    zero_division=0)
weight_f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

print()
print('=' * 55)
print('GRU — TEST SET RESULTS')
print('=' * 55)
print(f'  Overall Accuracy : {test_acc:.4f}  ({test_acc*100:.1f}%)')
print(f'  Macro F1 Score   : {macro_f1:.4f}  ({macro_f1*100:.1f}%)')
print(f'  Weighted F1 Score: {weight_f1:.4f}  ({weight_f1*100:.1f}%)')
print()

acc_ok = '✅ PASS' if test_acc >= 0.85 else '❌ BELOW TARGET'
f1_ok  = '✅ PASS' if macro_f1 >= 0.80 else '❌ BELOW TARGET'
print(f'  Target accuracy > 85% : {acc_ok}  ({test_acc*100:.1f}%)')
print(f'  Target macro F1 > 80% : {f1_ok}   ({macro_f1*100:.1f}%)')

# Generalized low-support signal diagnostic (Hesitant AND Aggressive, not
# just Hesitant) -- is there real signal even where argmax loses?
def print_signal_diagnostic(X, y, label, tag=''):
    probs = best_model.predict(X, verbose=0)
    mask = (y == label)
    if mask.sum() == 0:
        print(f'  [{tag}] no examples of {LABEL_NAMES[label]} in this set -- skip')
        return
    p_true = probs[mask, label].mean()
    p_other = probs[~mask, label].mean()
    print(f'  [{tag}] {LABEL_NAMES[label]:<12} mean P(class) on true examples: {p_true:.4f}  |  '
          f'on everything else: {p_other:.4f}  |  gap: {p_true - p_other:+.4f}')

print()
print('Signal diagnostic for low-support classes (is there real signal, even where argmax loses?):')
for lbl in range(N_CLASSES):
    if int(np.sum(y_test == lbl)) < LOW_SUPPORT_THRESHOLD or int(np.sum(y_val == lbl)) < LOW_SUPPORT_THRESHOLD:
        print_signal_diagnostic(X_test, y_test, lbl, tag='test')
        print_signal_diagnostic(X_val, y_val, lbl, tag='val')

print()
print('Per-class Classification Report:')
print()
report = classification_report(y_test, y_pred, target_names=LABEL_NAMES, zero_division=0)
print(report)

report_dict = classification_report(y_test, y_pred, target_names=LABEL_NAMES, zero_division=0, output_dict=True)
with open(f'{MODEL_PATH}/gru_classification_report.json', 'w') as f:
    json.dump(report_dict, f, indent=2)
print(f'Saved: {MODEL_PATH}/gru_classification_report.json')

# ============================================================
# 7. CONFUSION MATRIX
# ============================================================

cm      = confusion_matrix(y_test, y_pred)
cm_norm = cm.astype('float') / cm.sum(axis=1, keepdims=True)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.heatmap(cm, annot=True, fmt='d', xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES,
            cmap='Blues', ax=axes[0])
axes[0].set_title('Confusion Matrix — Raw Counts', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('True Label')
axes[0].tick_params(axis='x', rotation=30)

sns.heatmap(cm_norm, annot=True, fmt='.2f', xticklabels=LABEL_NAMES, yticklabels=LABEL_NAMES,
            cmap='Blues', vmin=0, vmax=1, ax=axes[1])
axes[1].set_title('Confusion Matrix — Normalised (Recall)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Predicted Label')
axes[1].set_ylabel('True Label')
axes[1].tick_params(axis='x', rotation=30)

plt.suptitle('GRU Mental State Classifier — Confusion Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(f'{MODEL_PATH}/gru_confusion_matrix.png', dpi=150, bbox_inches='tight')
plt.show()
print(f'Saved: {MODEL_PATH}/gru_confusion_matrix.png')

# ============================================================
# 8. PER-CLASS F1
# ============================================================

per_class_f1 = f1_score(y_test, y_pred, average=None, zero_division=0)
colors = ['#22C55E' if f >= 0.80 else '#F59E0B' if f >= 0.60 else '#EF4444' for f in per_class_f1]

plt.figure(figsize=(10, 5))
bars = plt.bar(LABEL_NAMES, per_class_f1, color=colors, edgecolor='white', linewidth=0.5)
plt.axhline(y=0.80, color='green',  linestyle='--', alpha=0.7, label='Target F1 = 0.80')
plt.axhline(y=0.60, color='orange', linestyle='--', alpha=0.5, label='Acceptable F1 = 0.60')
for bar, f1 in zip(bars, per_class_f1):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.01, f'{f1:.3f}',
              ha='center', va='bottom', fontsize=11, fontweight='bold')
plt.title('GRU — F1 Score per Mental State Class', fontsize=14, fontweight='bold')
plt.xlabel('Mental State')
plt.ylabel('F1 Score')
plt.ylim([0, 1.1])
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(f'{MODEL_PATH}/gru_per_class_f1.png', dpi=150, bbox_inches='tight')
plt.show()

print('Per-class F1 scores:')
for name, f1 in zip(LABEL_NAMES, per_class_f1):
    status = '✅' if f1 >= 0.80 else '⚠️' if f1 >= 0.60 else '❌'
    print(f'  {status} {name:<15}  F1 = {f1:.4f}')

# ============================================================
# 9. SAVE
# ============================================================

best_model.export(f'{MODEL_PATH}/gru_mental_state_saved')

summary = {
    'architecture': {
        'input_shape': [SEQ_LEN, N_FEATURES],
        'gru_1_units': 128, 'dropout_1': 0.3,
        'gru_2_units': 64, 'dropout_2': 0.3,
        'output_units': N_CLASSES, 'activation': 'softmax',
        'total_params': int(best_model.count_params())
    },
    'training': {
        'epochs_run': len(history.history['loss']),
        'batch_size': BATCH_SIZE,
        'optimizer': 'Adam',
        'learning_rate': 1e-3,
        'loss': 'sparse_categorical_crossentropy',
        'class_weighting': f'effective_number_of_samples (beta={EFFECTIVE_NUM_BETA})',
        'class_weights': class_weights,
        'checkpoint_metric': f'val_macro_f1 (smoothed, window={F1_SMOOTHING_WINDOW})',
        'best_smoothed_val_macro_f1': float(f1_checkpoint.best_smoothed),
        'best_raw_val_macro_f1': float(f1_checkpoint.best_raw_f1),
        'best_epoch': f1_checkpoint.best_epoch
    },
    'test_results': {
        'accuracy': float(test_acc),
        'macro_f1': float(macro_f1),
        'weighted_f1': float(weight_f1),
        'per_class_f1': {name: float(f1) for name, f1 in zip(LABEL_NAMES, per_class_f1)}
    },
    'targets_met': {
        'accuracy_85pct': bool(test_acc >= 0.85),
        'macro_f1_80pct': bool(macro_f1 >= 0.80)
    },
    'label_names': LABEL_NAMES
}

with open(f'{MODEL_PATH}/gru_training_summary.json', 'w') as f:
    json.dump(summary, f, indent=2)

print('=' * 55)
print('TRAINING COMPLETE — FINAL SUMMARY')
print('=' * 55)
print(f'  Epochs run        : {summary["training"]["epochs_run"]}')
print(f'  Best val macro F1 : {summary["training"]["best_smoothed_val_macro_f1"]:.4f}  @ epoch {summary["training"]["best_epoch"]}')
print(f'  Test accuracy     : {test_acc*100:.1f}%  (target >85%)')
print(f'  Test macro F1     : {macro_f1*100:.1f}%  (target >80%)')
print()
print('Saved files:')
for fname in [
    'gru_mental_state_best.keras', 'gru_training_curves.png', 'gru_confusion_matrix.png',
    'gru_per_class_f1.png', 'gru_classification_report.json', 'gru_training_summary.json', 'gru_training_log.csv'
]:
    fp = f'{MODEL_PATH}/{fname}'
    exists = os.path.exists(fp)
    size = f'{os.path.getsize(fp)/1024:.0f} KB' if exists else ''
    print(f'  {"✅" if exists else "❌"} {fname:<40} {size}')
```
