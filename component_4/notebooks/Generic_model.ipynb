```python
# ==============================
# 2. IMPORT LIBRARIES
# ==============================

from ultralytics import YOLO
import os
import shutil
```

```python
# ==============================
# 2. MOUNT GOOGLE DRIVE
# ==============================

from google.colab import drive
drive.mount('/content/drive')
```

```python
# ==============================
# 3. CHECK DATASET YAML PATH
# ==============================
# This YAML was created in data_preparation.ipynb.
# It has only one class:
# 0: traffic_sign

yaml_path = "/content/drive/MyDrive/research/yolo_datasets/generic_traffic_sign/generic_traffic_sign.yaml"

print("YAML exists:", os.path.exists(yaml_path))
```

```python
# ==============================
# 4. TRAIN GENERIC TRAFFIC SIGN DETECTOR
# ==============================
# This model only detects whether a traffic sign exists.
# It does NOT identify sign type.

generic_model = YOLO("yolov8n.pt")

generic_model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    name="generic_traffic_sign_detector"
)
```

```python
# ==============================
# 5. SAVE TRAINED MODEL TO DRIVE
# ==============================
# Save best.pt so we can use it later in Vehicle_A.ipynb.

save_dir = "/content/drive/MyDrive/research/trained_models"
os.makedirs(save_dir, exist_ok=True)

src = "/content/runs/detect/generic_traffic_sign_detector/weights/best.pt"
dst = f"{save_dir}/generic_traffic_sign_detector_best.pt"

shutil.copy(src, dst)

print("Saved:", dst)
```
